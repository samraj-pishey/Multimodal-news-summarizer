# CLIP-selected image captions as extra input to BART (MM-AVS)

Experiment A of the CLIP stage. Every variant fine-tunes `facebook/bart-large-cnn` with the **same cleaning,
hyperparameters, generation settings and validation-based epoch selection as run 3**; only the input changes.

Variants (fixed before any result was seen):

| Variant | Input to BART |
|---|---|
| `article` | article only (re-run of the run 3 baseline; also measures run-to-run noise) |
| `all_captions` | all image captions + article |
| `clip_top2` | captions of the 2 images CLIP ViT-B/32 rates most similar to the article (title + lead) + article |
| `random2` | 2 randomly chosen captions + article (control for "fewer captions") |

Captions come first so that article truncation at 1,024 tokens never removes them. CNN stories have no
images, so their input is identical in every variant. Caption selection was done offline
(`src/clip_scores.py`, `src/make_caption_inputs.py`) and is read from `caption_inputs.jsonl`.

**Before running:** attach dataset **mmavs-text** (version with `caption_inputs.jsonl`), GPU T4, Internet on.
Run once with `PILOT = True`, then `PILOT = False` and *Save Version → Save & Run All*.

```python
import glob, json, math, os, random, re, time
from collections import Counter
from pathlib import Path

import numpy as np
import torch
import transformers
from torch.utils.data import DataLoader
from transformers import AutoModelForSeq2SeqLM, AutoTokenizer, get_linear_schedule_with_warmup
from rouge_score import rouge_scorer

# ---- configuration -------------------------------------------------------
PILOT = False                 # True: tiny smoke test. False: full run.

MODEL_NAME = "facebook/bart-large-cnn"
SEED = 42
MAX_SOURCE_LEN = 1024         # BART's limit; ~25% of articles are longer and get truncated
MAX_TARGET_LEN = 192          # covers ~99% of reference summaries
NUM_EPOCHS = 4                # run 3 selected epoch 3 of 6; later epochs over-fit
LEARNING_RATE = 2e-5
WEIGHT_DECAY = 0.01
WARMUP_RATIO = 0.1
BATCH_SIZE = 2
GRAD_ACCUM = 4                # effective batch size 8
GRADIENT_CHECKPOINTING = True # trades ~30% speed for memory headroom on a 16 GB T4
TRAIN_ON_REPS_ONLY = True     # one copy per duplicate story group

# Fixed before seeing any test result; identical for pretrained and fine-tuned models.
GEN_KWARGS = dict(num_beams=4, no_repeat_ngram_size=3, length_penalty=2.0,
                  min_new_tokens=20, max_new_tokens=160, early_stopping=True)
EVAL_BATCH_SIZE = 8

OUT_DIR = Path("/kaggle/working")
VARIANTS = ["article", "all_captions", "clip_top2", "random2"]

if PILOT:
    NUM_EPOCHS = 1

def set_seed(seed):
    random.seed(seed); np.random.seed(seed); torch.manual_seed(seed); torch.cuda.manual_seed_all(seed)

set_seed(SEED)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device, torch.cuda.get_device_name(0) if device.type == "cuda" else "")
print("torch", torch.__version__, "| transformers", transformers.__version__)
```

## Load the splits and captions

```python
data_dir = Path(glob.glob("/kaggle/input/**/train.jsonl", recursive=True)[0]).parent
print("data dir:", data_dir)

raw = {split: [json.loads(line) for line in open(data_dir / f"{split}.jsonl")]
       for split in ("train", "val", "test")}

# MM-AVS Daily Mail highlights are glued together with no separator at all
# ("...last FridayRaiders seen on CCTV..."). Split where a lowercase letter, digit
# or closing quote runs into a capitalised word, unless the boundary falls inside a
# camelCase word that also appears in the (correctly spaced) articles, e.g.
# iPhone, YouTube, DailyMailTV, or a Mc/Mac surname.
REAL_CAMEL = {w for w, c in Counter(
    w for rows in raw.values() for r in rows
    for w in re.findall(r"[A-Za-z]*[a-z][A-Z][A-Za-z]*", r["article"])).items() if c >= 2}
# A quote only counts as a highlight end when it closes a word ("...said'Police").
GLUED = re.compile(r"(?:(?<=[a-z0-9%)])|(?<=[\w,]['\u2019\"]))(?=[A-Z][a-z]|[AI]\b|[A-Z]{2,}\b)")
GLUED_STOP = re.compile(r"(?:(?<=[a-z0-9%)'\u2019\" ][.!?]))(?=[A-Z\"'\u2018])")   # "...one.Parental", "...one .Parental"

def inside_real_word(text, pos):
    a, b = pos, pos
    while a > 0 and text[a - 1].isalpha():
        a -= 1
    while b < len(text) and text[b].isalpha():
        b += 1
    run, k = text[a:b], pos - a
    if run[max(0, k - 3):k] == "Mac" or run[max(0, k - 2):k] == "Mc":
        return True
    return any(run[i:j] in REAL_CAMEL for i in range(k) for j in range(k + 1, len(run) + 1))

def split_highlights(text):
    text = GLUED_STOP.sub("\n", text.replace("\xa0", " "))
    cuts = [m.start() for m in GLUED.finditer(text) if not inside_real_word(text, m.start())]
    for pos in reversed(cuts):
        text = text[:pos] + "\n" + text[pos:]
    parts = []
    for s in (s.strip() for s in text.split("\n")):
        if parts and re.fullmatch(r"[\'\u2019\"]+", s):
            parts[-1] += s          # closing quote split off its sentence ("...said.'")
        elif s:
            parts.append(s)
    return parts

def clean_summary(text):
    # One highlight per sentence, each ending in a full stop, so the model learns
    # well-formed sentences and rougeLsum can split them.
    highlights = [re.sub(r"\s+", " ", h) for h in split_highlights(text)]
    return " ".join(h if h[-1] in ".!?" else h + "." for h in highlights)

# Articles have the same scraping damage: sentences run together ("...the sideline.Feki was..."),
# non-breaking spaces, a glued "(CNN)" dateline and a few CNN boilerplate lines.
GLUED_SENTENCE = re.compile(r"(?<=[a-z0-9%)][.!?])(?=[A-Z\u2018\u201c])|(?<=[a-z0-9%)][.!?][\u2019\u201d'\"])(?=[A-Z])")
BOILERPLATE = re.compile(r"Copyright \d{4} The Conversation\.|Some rights reserved\.|"
                         r"Share your thoughts with [^.]{1,60} on Twitter @\w+\.|"
                         r"(?<=[.!?] )[A-Z][^.!?]{1,80} contributed to this report\.")

def clean_article(text):
    text = re.sub(r"^\s*\(CNN\)\s*(--)?\s*", "", text.replace("\xa0", " "))
    text = BOILERPLATE.sub("", GLUED_SENTENCE.sub(" ", text))
    return re.sub(r"\s+", " ", text).strip()

def load(split):
    rows = [dict(r) for r in raw[split]]
    for r in rows:
        r["article"] = clean_article(r["article"])
        r["summary"] = clean_summary(r["summary"])
    return rows

train_rows = load("train")
val_rows = [r for r in load("val") if r["is_group_rep"]]
test_rows = [r for r in load("test") if r["is_group_rep"]]   # each test story counted once
if TRAIN_ON_REPS_ONLY:
    train_rows = [r for r in train_rows if r["is_group_rep"]]

if PILOT:
    # Spread the pilot sample over both sources (rows are sorted CNN first), so captions are exercised.
    train_rows, val_rows, test_rows = train_rows[::28][:48], val_rows[::10][:16], test_rows[::26][:16]

print(f"train {len(train_rows)} | val {len(val_rows)} | test {len(test_rows)}")
print(json.load(open(data_dir / "splits_summary.json"))["counts"])
print("example cleaned reference:", test_rows[-1]["summary"][:300])

# Image captions selected offline with CLIP (see the introduction).
captions = {}
for line in open(data_dir / "caption_inputs.jsonl"):
    row = json.loads(line)
    captions[row["id"]] = row

def build_input(row, variant):
    caps = [] if variant == "article" else captions.get(row["id"], {}).get(variant, [])
    if not caps:
        return row["article"]
    return "Image captions: " + " | ".join(caps) + " || Article: " + row["article"]

def set_inputs(variant):
    for rows in (train_rows, val_rows, test_rows):
        for r in rows:
            r["input"] = build_input(r, variant)

print("stories with captions:", {s: sum(r["id"] in captions and bool(captions[r["id"]]["all_captions"]) for r in rows)
                                 for s, rows in (("train", train_rows), ("val", val_rows), ("test", test_rows))})
```

## Tokenize

```python
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

def encode(rows):
    enc = tokenizer([r["input"] for r in rows], max_length=MAX_SOURCE_LEN, truncation=True)
    labels = tokenizer(text_target=[r["summary"] for r in rows], max_length=MAX_TARGET_LEN, truncation=True)
    return [{"input_ids": enc["input_ids"][i], "labels": labels["input_ids"][i]} for i in range(len(rows))]

def truncation_report(rows, name):
    a = [len(x) for x in tokenizer([r["input"] for r in rows])["input_ids"]]
    s = [len(x) for x in tokenizer(text_target=[r["summary"] for r in rows])["input_ids"]]
    print(f"{name}: articles > {MAX_SOURCE_LEN} tokens: {np.mean(np.array(a) > MAX_SOURCE_LEN):.1%} | "
          f"summaries > {MAX_TARGET_LEN} tokens: {np.mean(np.array(s) > MAX_TARGET_LEN):.1%}")

def collate(batch):
    pad = tokenizer.pad_token_id
    src_len = max(len(b["input_ids"]) for b in batch)
    tgt_len = max(len(b["labels"]) for b in batch)
    input_ids = torch.full((len(batch), src_len), pad)
    attention = torch.zeros((len(batch), src_len), dtype=torch.long)
    labels = torch.full((len(batch), tgt_len), -100)   # -100 = ignored by the loss
    for i, b in enumerate(batch):
        input_ids[i, :len(b["input_ids"])] = torch.tensor(b["input_ids"])
        attention[i, :len(b["input_ids"])] = 1
        labels[i, :len(b["labels"])] = torch.tensor(b["labels"])
    return {"input_ids": input_ids, "attention_mask": attention, "labels": labels}
```

## Generation, ROUGE and significance helpers

```python
scorer = rouge_scorer.RougeScorer(["rouge1", "rouge2", "rougeL", "rougeLsum"], use_stemmer=True)
METRICS = ["rouge1", "rouge2", "rougeL", "rougeLsum"]

def sentences_per_line(text):
    # rougeLsum compares sentence by sentence; it expects one sentence per line.
    return "\n".join(s for s in re.split(r"(?<=[.!?])\s+", text.strip()) if s)

@torch.no_grad()
def generate(model, rows):
    model.eval()
    preds = []
    for i in range(0, len(rows), EVAL_BATCH_SIZE):
        batch = rows[i:i + EVAL_BATCH_SIZE]
        enc = tokenizer([r["input"] for r in batch], max_length=MAX_SOURCE_LEN,
                        truncation=True, padding=True, return_tensors="pt").to(device)
        with torch.autocast("cuda", dtype=torch.float16, enabled=device.type == "cuda"):
            out = model.generate(**enc, **GEN_KWARGS)
        preds += tokenizer.batch_decode(out, skip_special_tokens=True)
    return [p.strip() for p in preds]

def score(rows, preds):
    per_example = []
    for r, p in zip(rows, preds):
        s = scorer.score(sentences_per_line(r["summary"]), sentences_per_line(p))
        per_example.append({m: s[m].fmeasure for m in METRICS})
    return per_example

def summarize(per_example, n_boot=1000):
    # Mean F1 with a 95% bootstrap confidence interval over test stories.
    rng = np.random.default_rng(SEED)
    out = {}
    for m in METRICS:
        v = np.array([e[m] for e in per_example])
        boots = [rng.choice(v, len(v)).mean() for _ in range(n_boot)]
        out[m] = {"mean": float(v.mean()), "ci95": [float(np.percentile(boots, 2.5)), float(np.percentile(boots, 97.5))]}
    return out

def rouge_mean(summary):
    return float(np.mean([summary[m]["mean"] for m in ("rouge1", "rouge2", "rougeL")]))

def paired_bootstrap(a_scores, b_scores, n_boot=1000):
    # Fraction of resamples (of test stories) in which b has the higher mean, per metric.
    rng = np.random.default_rng(SEED)
    diffs = np.array([[b[m] - a[m] for m in METRICS] for a, b in zip(a_scores, b_scores)])
    wins = np.mean([diffs[rng.integers(0, len(diffs), len(diffs))].mean(0) > 0 for _ in range(n_boot)], axis=0)
    return {m: float(w) for m, w in zip(METRICS, wins)}
```

## Train and evaluate one variant

```python
def run_variant(variant):
    set_seed(SEED)
    set_inputs(variant)
    truncation_report(train_rows, f"{variant} train")
    train_data, val_data = encode(train_rows), encode(val_rows)
    g = torch.Generator().manual_seed(SEED)
    train_loader = DataLoader(train_data, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate, generator=g)
    val_loader = DataLoader(val_data, batch_size=BATCH_SIZE, collate_fn=collate)

    model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME).to(device)
    if GRADIENT_CHECKPOINTING:
        model.gradient_checkpointing_enable()
        model.config.use_cache = False
    no_decay = ["bias", "layer_norm.weight", "layernorm_embedding.weight"]
    params = [
        {"params": [p for n, p in model.named_parameters() if not any(k in n for k in no_decay)], "weight_decay": WEIGHT_DECAY},
        {"params": [p for n, p in model.named_parameters() if any(k in n for k in no_decay)], "weight_decay": 0.0},
    ]
    optimizer = torch.optim.AdamW(params, lr=LEARNING_RATE)
    total_steps = math.ceil(len(train_loader) / GRAD_ACCUM) * NUM_EPOCHS
    scheduler = get_linear_schedule_with_warmup(optimizer, int(WARMUP_RATIO * total_steps), total_steps)
    scaler = torch.amp.GradScaler(enabled=device.type == "cuda")
    best_dir = OUT_DIR / f"best-{variant}"

    history, best = [], {"rouge_mean": -1.0, "epoch": None}
    start = time.time()
    if device.type == "cuda":
        torch.cuda.reset_peak_memory_stats()
    for epoch in range(1, NUM_EPOCHS + 1):
        model.train()
        running, t0 = [], time.time()
        optimizer.zero_grad()
        for step, batch in enumerate(train_loader, 1):
            batch = {k: v.to(device) for k, v in batch.items()}
            with torch.autocast("cuda", dtype=torch.float16, enabled=device.type == "cuda"):
                loss = model(**batch).loss / GRAD_ACCUM
            scaler.scale(loss).backward()
            running.append(loss.item() * GRAD_ACCUM)
            if step % GRAD_ACCUM == 0 or step == len(train_loader):
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
                scaler.step(optimizer)
                scaler.update()
                scheduler.step()
                optimizer.zero_grad()

        model.config.use_cache = True
        model.eval()
        with torch.no_grad():
            losses = []
            for batch in val_loader:
                batch = {k: v.to(device) for k, v in batch.items()}
                with torch.autocast("cuda", dtype=torch.float16, enabled=device.type == "cuda"):
                    losses.append(model(**batch).loss.item())
        v_preds = generate(model, val_rows)
        v_summary = summarize(score(val_rows, v_preds), n_boot=1)
        if GRADIENT_CHECKPOINTING:
            model.config.use_cache = False

        record = {"epoch": epoch, "train_loss": float(np.mean(running)), "val_loss": float(np.mean(losses)),
                  "val_rouge": {m: v_summary[m]["mean"] for m in METRICS},
                  "val_rouge_mean": rouge_mean(v_summary), "minutes": (time.time() - t0) / 60}
        history.append(record)
        print(f"[{variant}] epoch {epoch}: train_loss {record['train_loss']:.4f} | val_loss {record['val_loss']:.4f} | "
              f"val mean R1/2/L {record['val_rouge_mean']:.4f} | {record['minutes']:.1f} min")
        if record["val_rouge_mean"] > best["rouge_mean"]:
            best = {"rouge_mean": record["val_rouge_mean"], "epoch": epoch}
            best_val_preds = v_preds
            model.save_pretrained(best_dir)

    train_minutes = (time.time() - start) / 60
    peak_gb = torch.cuda.max_memory_allocated() / 1e9 if device.type == "cuda" else None
    del model, optimizer
    torch.cuda.empty_cache()

    model = AutoModelForSeq2SeqLM.from_pretrained(best_dir).to(device)   # reload what was actually saved
    preds = generate(model, test_rows)
    scores = score(test_rows, preds)
    del model
    torch.cuda.empty_cache()
    shutil.rmtree(best_dir)          # keep the output small; predictions are what we analyse

    print(f"[{variant}] best epoch {best['epoch']} | training {train_minutes:.1f} min | test "
          + " ".join(f"{m} {np.mean([s[m] for s in scores]):.4f}" for m in METRICS))
    return {"variant": variant, "best_epoch": best["epoch"], "history": history,
            "training_minutes": train_minutes, "peak_gpu_memory_gb": peak_gb,
            "test_preds": preds, "test_scores": scores, "val_preds": best_val_preds,
            "test_summary": summarize(scores),
            "test_inputs_with_captions": sum(r["input"] != r["article"] for r in test_rows)}
```

## Run all variants

```python
import csv, shutil

def save_variant(out):
    v = out["variant"]
    with open(OUT_DIR / f"test_predictions_{v}.csv", "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["id", "source", "reference", "prediction"] + METRICS)
        for r, p, s in zip(test_rows, out["test_preds"], out["test_scores"]):
            w.writerow([r["id"], r["source"], r["summary"], p] + [round(s[m], 4) for m in METRICS])
    with open(OUT_DIR / f"val_predictions_{v}.csv", "w", newline="") as f:
        w = csv.writer(f)
        w.writerow(["id", "source", "reference", "prediction"])
        for r, p in zip(val_rows, out["val_preds"]):
            w.writerow([r["id"], r["source"], r["summary"], p])

runs = {}
for variant in VARIANTS:
    runs[variant] = run_variant(variant)
    save_variant(runs[variant])      # saved as we go, so a late failure keeps earlier variants
```

## Compare with the article-only variant

```python
dm = [i for i, r in enumerate(test_rows) if r["source"] == "dailymail"]
pick = lambda scores, idx: [scores[i] for i in idx]
base = runs["article"]["test_scores"]

comparison = {}
print(f"{'variant':<14}" + "".join(f"{m:>11}" for m in METRICS) + "   P(better than article), all / DM")
for v, out in runs.items():
    s = out["test_scores"]
    row = {"all": {m: float(np.mean([x[m] for x in s])) for m in METRICS},
           "dailymail": {m: float(np.mean([x[m] for x in pick(s, dm)])) for m in METRICS}}
    if v != "article":
        row["p_better_than_article_all"] = paired_bootstrap(base, s)
        row["p_better_than_article_dm"] = paired_bootstrap(pick(base, dm), pick(s, dm)) if dm else {}
    comparison[v] = row
    p = (" / ".join(f"{row['p_better_than_article_all'][m]:.2f}|{row['p_better_than_article_dm'].get(m, float('nan')):.2f}" for m in METRICS)
         if v != "article" else "")
    print(f"{v:<14}" + "".join(f"{row['all'][m]:>11.4f}" for m in METRICS) + "   " + p)

results = {
    "pilot": PILOT, "model": MODEL_NAME,
    "versions": {"torch": torch.__version__, "transformers": transformers.__version__},
    "gpu": torch.cuda.get_device_name(0) if device.type == "cuda" else "cpu",
    "data": {"train": len(train_rows), "val": len(val_rows), "test": len(test_rows), "test_dailymail": len(dm)},
    "hyperparameters": {"seed": SEED, "max_source_len": MAX_SOURCE_LEN, "max_target_len": MAX_TARGET_LEN,
                        "epochs": NUM_EPOCHS, "lr": LEARNING_RATE, "weight_decay": WEIGHT_DECAY,
                        "warmup_ratio": WARMUP_RATIO, "batch_size": BATCH_SIZE, "grad_accum": GRAD_ACCUM},
    "generation": GEN_KWARGS,
    "selection": "best epoch by mean of val ROUGE-1/2/L F1",
    "variants": {v: {k: out[k] for k in ("best_epoch", "history", "training_minutes", "peak_gpu_memory_gb",
                                         "test_summary", "test_inputs_with_captions")} for v, out in runs.items()},
    "comparison": comparison,
}
(OUT_DIR / "results.json").write_text(json.dumps(results, indent=2))
print("saved:", sorted(p.name for p in OUT_DIR.iterdir()))
```

## Notes
- A variant only counts as an improvement if it beats `article` by more than run-to-run noise
  (runs 2 and 3 of the same setup differed by up to 0.6 ROUGE points) **and** the paired bootstrap favours it.
- `random2` vs `clip_top2` isolates the value of CLIP's selection from the effect of adding two captions.
- Captions push some article text past the 1,024-token limit for long articles; `truncation_report` prints how often.
